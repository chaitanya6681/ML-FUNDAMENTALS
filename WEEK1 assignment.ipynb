```python
#Loading dataset and summarizing basic stats
import pandas as pd
d=pd.read_csv('/kaggle/input/datasets/yasserh/housing-prices-dataset/Housing.csv')
d.info()
d.describe()
d.head(10)
```

```python
n_cols=d.select_dtypes(include=['int64','float64']).columns
for col in n_cols:
    d[col]=d[col].fillna(d[col].median())
d.head(10)
```

```python
#Label Encoding
from sklearn.preprocessing import LabelEncoder
l=LabelEncoder()
c_cols=d.select_dtypes(include='object').columns
for col in c_cols:
    d[col]=l.fit_transform(d[col].astype(str))
d.head(10)
```

```python
#One-Hot Encoding
d=pd.get_dummies(d,columns=c_cols,drop_first=True)
print(d.shape)
```
